### **Problem-1:** Make a subplots which have 2 plots.

- For the first chart, draw a scatter plot "*Monitored Cap.(MW)*" vs "*Total Cap. Under Maintenance (MW)*" of top 5 most frequent power stations. Then draw the lines which indicate the average values of these two columns. Change the colors according to the names of the Power Stations.
- For the second chart, draw a scatter plot "*Monitored Cap. (MW)*" vs "*Actual(MU)*" of the top 5 most frequent power stations. Also draw the lines which indicates the average values of these two columns. Change the colors according to the names of the Power Stations.


````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/PowerGeneration - PowerGeneration.csv")

top5 = df['Power Station'].value_counts().head(5).index

top_df = df[df['Power Station'].isin(top5)]

avg_monitored_1 = top_df['Monitored Cap.(MW)'].mean()
avg_maintenance = top_df['Total Cap. Under Maintenace (MW)'].mean()

avg_monitored_2 = top_df['Monitored Cap.(MW)'].mean()
avg_actual = top_df['Actual(MU)'].mean()

fig, ax = plt.subplots(1,2,figsize=(18,7))

for station in top5:

    station_df = top_df[top_df['Power Station'] == station]

    ax[0].scatter(station_df['Monitored Cap.(MW)'],
                  station_df['Total Cap. Under Maintenace (MW)'],
                  label=station,
                  alpha=0.7,
                  s=80)

ax[0].axvline(avg_monitored_1,color='red',linestyle='--',linewidth=2,label='Avg Monitored Cap.')
ax[0].axhline(avg_maintenance,color='blue',linestyle='--',linewidth=2,label='Avg Maintenance Cap.')

ax[0].set_xlabel('Monitored Cap. (MW)')
ax[0].set_ylabel('Total Cap. Under Maintenace (MW)')
ax[0].set_title('Monitored Capacity vs Maintenance Capacity')
ax[0].grid(True)
ax[0].legend()

for station in top5:

    station_df = top_df[top_df['Power Station'] == station]

    ax[1].scatter(station_df['Monitored Cap.(MW)'],
                  station_df['Actual(MU)'],
                  label=station,
                  alpha=0.7,
                  s=80)

ax[1].axvline(avg_monitored_2,color='red',linestyle='--',linewidth=2,label='Avg Monitored Cap.')
ax[1].axhline(avg_actual,color='green',linestyle='--',linewidth=2,label='Avg Actual(MU)')

ax[1].set_xlabel('Monitored Cap. (MW)')
ax[1].set_ylabel('Actual(MU)')
ax[1].set_title('Monitored Capacity vs Actual Generation')
ax[1].grid(True)
ax[1].legend()

plt.suptitle('Power Station Analysis',fontsize=18)

plt.tight_layout()

plt.show()
````

### **Problem-3:** Make a 3D *Surface* plot of this below mathematical equation.

$$z = |x| - |y|$$

````python
# code here
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

X,Y = np.meshgrid(x,y)

Z = np.abs(X) - np.abs(Y)

fig = plt.figure(figsize=(12,8))

ax = fig.add_subplot(111,projection='3d')

surface = ax.plot_surface(X,Y,Z,cmap='viridis')

ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')

ax.set_title('3D Surface Plot of z = |x| - |y|')

fig.colorbar(surface,shrink=0.5,label='Z Value')

plt.show()
````

### `Problem-6` Use Pandas plot functions

* Line plot of closing value of top 5 Stocks in Year 2020.
* Take top 5 stocks based on total turnover in Year 2020

````python
# code here
df = pd.read_csv('/content/nifty-50 - nifty-50.csv')

df.head()

df['Date'] = pd.to_datetime(df['Date'])

df_2020 = df[df['Date'].dt.year == 2020]

top5 = df_2020.groupby('Symbol')['Turnover'].sum().sort_values(ascending=False).head(5).index

top_df = df_2020[df_2020['Symbol'].isin(top5)]

pivot_df = top_df.pivot_table(index='Date',
                              columns='Symbol',
                              values='Close')

pivot_df.plot(figsize=(15,7),
              linewidth=2)

plt.xlabel('Date')
plt.ylabel('Closing Price')

plt.title('Closing Value of Top 5 Stocks in Year 2020')

plt.grid(True)

plt.legend(title='Stocks')

plt.show()
````

### `Problem-7` Scatter plot Close price vs Volume for TOP-5 Stocks in year 2021
* Color on Symbol Column

````python
# code here
df['Date'] = pd.to_datetime(df['Date'])

df_2021 = df[df['Date'].dt.year == 2021]

top5 = df_2021.groupby('Symbol')['Turnover'].sum().sort_values(ascending=False).head(5).index

top_df = df_2021[df_2021['Symbol'].isin(top5)]

plt.figure(figsize=(15,7))

for stock in top5:

    stock_df = top_df[top_df['Symbol'] == stock]

    plt.scatter(stock_df['Close'],
                stock_df['Volume'],
                label=stock,
                alpha=0.7,
                s=50)

plt.xlabel('Close Price')
plt.ylabel('Volume')

plt.title('Close Price vs Volume for Top 5 Stocks in Year 2021')

plt.grid(True)

plt.legend(title='Stocks')

plt.show()
````

### `Problem-8` Create a 3-D Scatter Plot using `time,x,y` on below synthetic data. and give color gradiant on `z`
```
# Create a 3D dataset
time = np.linspace(0, 10, 100)
x = np.sin(time)
y = np.cos(time)
z = time

# Create a DataFrame from the dataset
data = pd.DataFrame({'time': time, 'x': x, 'y': y, 'z': z})
```

````python
# code here
time = np.linspace(0,10,100)

x = np.sin(time)
y = np.cos(time)
z = time

data = pd.DataFrame({'time':time,
                     'x':x,
                     'y':y,
                     'z':z})

fig = plt.figure(figsize=(12,8))

ax = fig.add_subplot(111,projection='3d')

scatter = ax.scatter(data['x'],
                     data['y'],
                     data['z'],
                     c=data['z'],
                     cmap='viridis',
                     s=60)

ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')

ax.set_title('3D Scatter Plot with Color Gradient on Z')

fig.colorbar(scatter,label='Z Value')

plt.show()
````

### `Problem 9:` Create a surface plot and the 2 types of the contour plots of the below equation.

$$z = sin(\sqrt{x^2 + y^2})$$


````python
# code here
x = np.linspace(-5,5,100)
y = np.linspace(-5,5,100)

X,Y = np.meshgrid(x,y)

Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure(figsize=(12,8))

ax = fig.add_subplot(131,projection='3d')

surface = ax.plot_surface(X,Y,Z,cmap='viridis')

ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')

ax.set_title('3D Surface Plot')

ax2 = fig.add_subplot(132)

contour1 = ax2.contour(X,Y,Z,20,cmap='plasma')

ax2.set_xlabel('X Axis')
ax2.set_ylabel('Y Axis')

ax2.set_title('Contour Plot')

fig.colorbar(contour1,ax=ax2)

ax3 = fig.add_subplot(133)

contour2 = ax3.contourf(X,Y,Z,20,cmap='coolwarm')

ax3.set_xlabel('X Axis')
ax3.set_ylabel('Y Axis')

ax3.set_title('Filled Contour Plot')

fig.colorbar(contour2,ax=ax3)

plt.suptitle('Surface and Contour Plots of z = sin(sqrt(x² + y²))',fontsize=16)

plt.tight_layout()

plt.show()
````

### `Problem 10:` Create a surface plot and the 2 types of the contour plots of the below equation.
